```python
#  Imports
import os

from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

load_dotenv()
```

```python
#  Create Gemini LLM
llm = ChatOpenAI(
    model=os.getenv("GPT_MODEL", "your-gpt-model"),
    base_url=os.getenv("PORTKEY_BASE_URL", "your-base-url"),
    api_key=os.getenv("PORTKEY_API_KEY", "your-api-key"),
)
print(llm.invoke("Say hi in five words.").content)
```

```python
#  Prompt to get user's name
name_prompt = PromptTemplate.from_template(
    "My name is {name}. Please remember it and greet me."
)
```

```python
#  Prompt to ask if it remembers the name
followup_prompt = PromptTemplate.from_template("Do you remember my name?")
```

```python
#  Create Chains
name_chain = name_prompt | llm | StrOutputParser()
followup_chain = followup_prompt | llm | StrOutputParser()
```

```python
#  Run both chains
name = "Savvas"  # input("Enter your name: ")
greet = name_chain.invoke({"name": name})  # First prompt - gives name
# print(greet)
print("\n  Greet:\n", greet)

print("\n Asking if it remembers your name...")
response = followup_chain.invoke({})  # Second prompt - ask if it remembers

print("\n  Response:\n", response)
```
